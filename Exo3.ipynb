```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt

t=np.linspace(-4,4,200)

porte1 = mana.porte((1/4)*t)

df = pd.DataFrame({"t":t,"porte1":porte1})

fig = px.line(df, x="t", y=["porte1"], title = "Courbe dans le même repère")
fig.show()
```

on calcule la transformée de fourier de f(t) = porte de t/4

La transformée de fourier est donc : 4 sinus cardinal de 4 nu



```python
nu = np.linspace(-10,10,200)

F = 4*np.sin(4*np.pi*nu)/(4*np.pi*nu)

df = pd.DataFrame({"nu":nu,"F":F})
fig = px.line(df, x="nu", y=["F"], title = "Courbe")
fig.show()
```

```python
I=np.trapezoid(abs(F)**2,nu)

print (I)

res = (4-I)/4

print(res)
```

Donc la précision relative du résultat obtenu est 0.025

ce résultat est obtenu à partir de la propriété de Rayleigh

Ce résultat dépend de la surface sous la porte à la base.
