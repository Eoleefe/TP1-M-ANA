```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt


k1 = 1
k2 = 3
k3 = 6
k4 = 9

t = np.linspace(-1,1,200)

f1 = k1*mana.porte(k1*t)
f2 = k2*mana.porte(k2*t)    
f3 = k3*mana.porte(k3*t)
f4 = k4*mana.porte(k4*t)

df = pd.DataFrame({"t":t,"f1":f1,"f2":f2,"f3":f3,"f4":f4})

fig = px.line(df, x="t  ", y=["f1","f2","f3","f4"], title = "Courbe")
fig.show()
```

Pour calculer les produits de convolution, on utlise numpy.convolve

```python
t = np.linspace(-4,4,200)

courbe = np.exp(-np.pi*t*t)*np.cos(4*np.pi*t)


fg1 = np.convolve(f1, courbe, mode='same')
fg2 = np.convolve(f2, courbe, mode='same')
fg3 = np.convolve(f3, courbe, mode='same')
fg4 = np.convolve(f4, courbe, mode='same')

df = pd.DataFrame({"t":t,"fg1":fg1,"fg2":fg2,"fg3":fg3,"fg4":fg4})

fig = px.line(df, x="t", y=["fg1","fg2","fg3","fg4"], title = "Courbe")
fig.show()
```
