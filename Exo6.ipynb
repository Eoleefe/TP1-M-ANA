```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt

t = np.linspace(-2, 2, 200)
dt = t[1] - t[0]

p1 = mana.porte(t)
p2 = mana.porte(t)

f = np.convolve(p1, p2, 'same') * dt
plt.plot(t, f)
```

```python

t = np.linspace(-1, 1, 200)
dt = t[1] - t[0]

p1 = mana.porte(t)
p2 = mana.porte(t)

p1_fourier, nu = mana.trans_fourier(p1, t)
p2_fourier, nu = mana.trans_fourier(p2, t)

f = np.convolve(p1, p2, 'same') * dt
df = pd.DataFrame({'t': t, 'f': f})

fig1  = px.line(df, x='t', y='f', title='Convolution de 2 portes')
fig1.show()

g = np.array(p1_fourier) * np.array(p2_fourier)

g2, t = mana.trans_fourier_inv(g, nu)

Re = np.real(g2)

dg = pd.DataFrame({'t': t, 'g2': Re})
fig2  = px.line(dg, x='t', y='g2', title='Inverse de la transformée de Fourier du produit(Re)')
fig2.show()

```
