```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
import scipy
from scipy.fft import fft2, ifft2, fftshift

def disque(n, m, r):
    x, y = np.meshgrid(np.arange(-n//2, n//2), np.arange(-m//2, m//2))
    D=(x**2+y**2<=r**2)
    return D

# lecture
A = plt.imread('barbara.jpg')
a, b = A.shape

# affichage de l'image originale
plt.figure(1,figsize=(9,9))
plt.subplot(2, 2, 1)
plt.imshow(A, cmap='gray')
plt.text(-50, 50, 'a', fontsize=14)
plt.axis('equal')
```

calcul de la transformée de fourier associée à Barbara et afficher son module

```python
f = fft2(A)
F = fftshift(f)

g = F
plt.figure(1,figsize=(9,9))
plt.subplot(2, 2, 1)
plt.imshow(np.log10(abs(g)), cmap='gray')
plt.text(-50, 50, 'a', fontsize=14)
plt.axis('equal')
```

On va filtrer avec un disque

```python
disque = disque(a, b, 140)

F = g*disque

plt.figure(1,figsize=(9,9))
plt.subplot(2, 2, 1)
plt.imshow(abs(np.log10(F)), cmap='gray')
plt.text(-50, 50, 'a', fontsize=14)
plt.axis('equal')
```

```python
F = ifft2(F)
plt.figure(1,figsize=(9,9))
plt.subplot(2, 2, 1)
plt.imshow(abs(np.log10(F)), cmap='gray')
plt.text(-50, 50, 'a', fontsize=14)
plt.axis('equal')
```
