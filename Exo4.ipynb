```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt

t = np.linspace(-5,5,200)

courbe = np.exp(-np.pi*t*t)*np.cos(4*np.pi*t)

df = pd.DataFrame({"t":t,"courbe":courbe})

fig = px.line(df, x="t", y=["courbe"], title = "Courbe dans le même repère")
fig.show()
```

on voit que la courbe extérieure à [-1, 1] est  nulle et à l'intérieur, elle est paire, avec des oscillations qui font une enveloppe qui suit presque un cercle

Transformée de fourier de la focntion f(t)

F(nu) = 1/2 * [exp(-pi(f-2)(f-2))+exp(-pi(f+2)(f+2))]

```python

F, nu = mana.trans_fourier(courbe, t)

Re = np.real(F)
Im = np.imag(F)
Phase = np.angle(F)

df = pd.DataFrame({"nu":nu,"Re":Re,"Im":Im})

fig = px.line(df, x="nu", y=["Re"], title = "Partie réelle et imaginaire de F")
fig.show()
```

```python
fig = px.line(df, x="nu", y=["Im"], title = "Partie réelle et imaginaire de F")
fig.show()
```
