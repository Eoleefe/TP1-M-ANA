```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
t=np.linspace(-4,4,200)
x=np.sin(2*np.pi*t)
y=mana.gauss(t)

df=pd.DataFrame({"t":t,"sinus":x,"gaussienne":y})


fig = px.line(df, x="t", y=["sinus","gaussienne"], title = "Deux courbes dans le même repère")
fig.show()
```
