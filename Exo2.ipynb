```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
```

```python
t = np.linspace(0,2,500)

courbe = np.exp(-t)

df=pd.DataFrame({"t":t,"courbe":courbe})

fig = px.line(df, x="t", y="courbe", title = "Courbe exponentielle décroissante")
fig.show()

N = 100

E_tot = np.trapezoid(courbe*courbe, t)/2

print("L'énergie totale du signal est  :  = ", E_tot)

a0 = 0.5*np.trapezoid(courbe, t)

E_rec = a0**2

s = a0

for k in range(1,N+1):
    
    ak = np.trapezoid(courbe*np.cos(np.pi*k*t))

    bk = np.trapezoid(courbe*np.sin(np.pi*k*t))

    s += (ak*np.cos(np.pi*k*t) + bk*np.sin(np.pi*k*t))

    E_rec += 0.5*(ak**2 + bk**2)

df=pd.DataFrame({"t":t,"courbe":courbe, "s":s})
    
fig = px.line(df, x="t", y=["courbe", "s"], title = "Courbe exponentielle décroissante")
fig.show()

```
