```python
import numpy as np
import msicpe.mana as mana
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
t = np.linspace(-2, 2, 200)

f = np.exp(-np.pi*(t**2))*np.cos(10*np.pi*t)

df = pd.DataFrame({"t":t, "f":f})

fig = px.line(df, t, ["f"], title = 'courbe expo')
fig.show()
```

```python
t = np.linspace(-10, 10, 200)

Y, nu = mana.trans_fourier(f, t)

Re = np.real(Y)
df = pd.DataFrame({"t":t, "Re":Re})

fig = px.line(df, t, ["Re"], title = 'courbe de la transfo de fourier')
fig.show()
```
